```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
import smtplib
import time
import datetime
import csv
```

```python
URL = 'https://www.amazon.com/Funny-Data-Systems-Business-Analyst/dp/B07FNW9FGJ/ref=sr_1_3?dchild=1&keywords=data%2Banalyst%2Btshirt&qid=1626655184&sr=8-3&customId=B0752XJYNL&th=1'

headers = {
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8,application/signed-exchange;v=b3;q=0.7",
    "Accept-Encoding": "gzip, deflate, br, zstd",
    "Accept-Language": "en-GB,en;q=0.9,en-US;q=0.8,fr-FR;q=0.7,fr;q=0.6",
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36 Edg/138.0.0.0"
  }

page = requests.get(URL, headers = headers)

Soup1 = BeautifulSoup(page.content, 'html.parser')
print(Soup1.prettify())
```

```python
title = Soup1.find(id="productTitle").get_text().strip()
print(title)

price_whole = Soup1.find(class_='a-price-whole').get_text(strip=True)
price_fraction = Soup1.find(class_='a-price-fraction').get_text(strip=True)
price = f'{price_whole}{price_fraction}'
print(price)
```

```python
today = datetime.date.today()
print(today)
```

```python
header = ['Title', 'Price','Date']
data = [title, price, today]

with open("AmazonWebScraperDataset.csv", "w", newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerow(data)
```
